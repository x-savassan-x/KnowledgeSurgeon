# Visualization

Reads scored `perf/`, `time/`, and `mem/` artifacts produced by the evaluation workflow.

## Environment

```python
import sys
from pathlib import Path

import yaml


def find_project_root(start: Path | None = None) -> Path:
    start = (start or Path.cwd()).resolve()
    for path in (start, *start.parents):
        if (path / "src").exists() and (path / "notebook").exists():
            return path
    raise RuntimeError(f"Could not find project root from {start}")


PROJECT_ROOT = find_project_root()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))
```

## Evaluation Config

```python
EVAL_CONFIG_PATH = PROJECT_ROOT / "src" / "evaluate" / "config.yaml"
with open(EVAL_CONFIG_PATH, encoding="utf-8") as file:
    eval_config = yaml.safe_load(file)

methods = list(eval_config["editing_methods"])
model = eval_config["model_name"]
milestones = list(eval_config["eval_milestones"])
n_runs = int(eval_config["n_runs"])
judge_repeats = int(eval_config["judge"]["repeats"])

# Add "base" here after the scoring worker materializes run-specific base perf files.
# methods = ["base", *methods]
```
